# A History of Programming

A history of programming is the story of how humans bridged the gap between human thought and the silent switches of digital hardware ([A Brief Introduction to Programming](https://youtu.be/9uW6B9LPntY)). It evolved from rewiring physical machines with patch cords into crafting abstract, expressive languages that run our modern world.


## The Pre-Digital Foundations: Mechanical Logic

Long before electronic computers, the fundamental concept of programming was born in the textile industry and mathematical theory.

* **1801 — The Jacquard Loom:** Joseph Marie Jacquard invented a loom controlled by **punched paper cards**. The sequence of holes dictated the pattern woven into fabric. This is widely considered the first instance of read-only programmed instructions.
* **1843 — The First Algorithm:** Mathematician **Ada Lovelace** translated an Italian article on Charles Babbage’s mechanical computer, the Analytical Engine. She added her own extensive notes, including an algorithm to calculate Bernoulli numbers. For this sequence of instructions designed for a machine, she is recognized as the world's first programmer.


## The Electronic Era: Hardware to Assembly (1940s–1950s)

Early 20th-century electronic computers were programmed directly by plugging cables into ([an example of a patch panel](https://share.google/WLfYbA4STPmoxvFds)) or manually flipping switches.

* **1945 — Stored-Program Architecture:** John von Neumann formalized the concept that a computer's instructions and data could reside in the same memory. Programs could now be loaded electronically.
* **1949 — Short Code & Assembly:** Programmers wrote instructions using raw machine code (binary). To make this easier, **Assembly language** was introduced, replacing binary opcodes with short text mnemonics (like `ADD` or `MOV`). Around the same time, John Mauchly proposed **Short Code**, one of the earliest high-level expression languages.


## The High-Level Language Revolution (1950s–1960s)

Writing code in assembly was tedious and machine-dependent. The late 1950s saw the birth of languages designed to look more like human logic and mathematical notation.

* **1957 — FORTRAN (Formula Translation):** Developed by John Backus at IBM, FORTRAN was the first major commercial high-level language. It revolutionized scientific and engineering computation. Used heavily in aerospace engineering, national laboratories, and weather forecasting (crunching heavy numbers)
* **1959 — COBOL (Common Business-Oriented Language):** Designed under the guidance of Grace Hopper to be readable by non-technical business professionals using English-like syntax. It became the backbone of financial institutions and government mainframes for decades.
* **1958 — LISP (LISt Processor):** Created by John McCarthy, LISP introduced concepts like conditional expressions and garbage collection. It became the dominant language for early Artificial Intelligence research.
* **1964 — BASIC (Beginner's All-purpose Symbolic Instruction Code):** Created by John Kemeny and Thomas Kurtz to democratize coding, allowing college students outside of math and science majors to write software.


## Structured Programming & Modern Foundations (1970s–1980s)

As software grew more complex, projects suffered from "spaghetti code" (overuse of `GOTO` statements). The 1970s introduced order, modularity, and systems programming.

* **1972 — C:** Developed by Dennis Ritchie at Bell Labs for the Unix operating system. C provided low-level memory access while maintaining high-level readability. It birthed the syntax family that would influence C++, Java, C#, JavaScript, and Python.
* **1972 — Prolog & SQL:** Logic programming entered the mainstream with Prolog (for rule-based AI), while Edgar C. Codd's relational database model led to **SQL**, transforming how data is queried.
* **1980s — Object-Oriented Programming (OOP):** Languages like **Smalltalk** and **C++** (developed by Bjarne Stroustrup in 1983) popularized OOP, allowing developers to model software around "objects" containing data and behavior.


## The Web Boom & Scripting (1990s)

The rise of the World Wide Web and personal computing shifted programming from enterprise and academic mainframes to everyday consumer technology.

* **1991 — Python:** Created by Guido van Rossum, emphasizing code readability with its distinctive white-space indentation.
* **1995 — Java:** Released by Sun Microsystems with the mantra "Write Once, Run Anywhere," making it an enterprise standard.
* **1995 — JavaScript & PHP:** JavaScript was created by Brendan Eich in just 10 days to bring interactivity to web browsers, while PHP powered dynamic server-side web pages.


## The Modern Era: Cloud, AI, and Abstraction (2000s–Present)

Today, programming languages abstract away hardware management almost entirely, focusing instead on scaling, productivity, and automation.

* **2009–2012 — Modern Languages:** Google introduced **Go** for distributed cloud systems, and Apple created **Swift** for robust mobile development. Microsoft released **TypeScript** in 2012 to bring static typing to JavaScript.
* **The Present:** Software development is increasingly augmented by AI code generation assistants, while **Python** dominates modern data science, machine learning, and cloud infrastructure pipelines.

# An Introduction Python Programming

Python is one of the world's most popular, versatile, and accessible programming languages. Created by Guido van Rossum and first released in 1991, Python was designed with a core philosophy centered on code readability and simplicity. It allows developers to express concepts in fewer lines of code than languages like C or Java, bridging the gap between human thought and hardware execution.

## Core Characteristics of Python

* **High-Level and Interpreted**: Python abstracts away complex memory management (like manual pointer allocation) and runs code line-by-line via an interpreter, making debugging faster and more intuitive.
* **Dynamically Typed**: You don't need to explicitly declare a variable's data type (such as int or str) beforehand; Python infers it automatically at runtime.
* **Multi-Paradigm** : Python supports multiple programming styles, including procedural, object-oriented (OOP), and functional programming.

## Basic Syntax & Concepts
Python's syntax relies heavily on clean indentation (whitespace) rather than curly braces {} or semicolons ;, which forces structured, readable code formatting.

### 1. Variables and basic data types

```python
name = "Modiri Mokaila" # this is what you call a string, alias 'str'. These kind of variables are used to store text/words/names data.
age = 26 # this is what you call a float, alias 'float'. These kind of variables are used to store decimal numbers, and can be rounded off.
is_active = True # this is what you call a boolean, alias 'bool'. These kind of variables are used to store True or False values.
```

### 2. Control Flow (Conditionals & Loops)

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
else:
    print("Grade: Needs Improvement")

# For loop iteration
for item in ["Pandas", "NumPy", "Scikit-learn"]:
    print(f"Library: {item}")
```

### 3. Functions

```python
def calculate_velocity(distance, time):
    """Calculate average velocity given distance and time."""
    if time == 0:
        return 0
    return distance / time

speed = calculate_velocity(100, 2.5)
print(f"Velocity: {speed} m/s")
```

## Python Libraries and Modules

A library (often used interchangeably with module in Python) is a pre-written collection of code that you can import into your own program to perform specific tasks without writing the logic from scratch. Think of it like a toolbox: instead of forging your own hammer and screwdriver (writing the code), you just go to the toolbox (import the library) and take out the exact tool you need for the job.

### The Key Distinction: Library vs. Module
* **Module** : This is the basic building block. A module is a single Python file (ending in **.py*) that contains functions, classes, or variables.
* **Library**: This is a broader term. It is a collection of related modules grouped together to form a substantial body of code for a specific purpose (e.g., NumPy is a library containing multiple modules that manage mathematical functions).

## How to Use a Library in Python
To use a module or library, you must "import" it into your script using the **import* keyword.

### 1. Importing an Entire Library

```python
# Import the standard 'math' library (containing mathematical functions such as pi, sqrt, add, subtract, log, etc.), which comes with Python
import math

# Use the 'pi' constant from the math toolbox
result = math.pi
print(f"The value of Pi is: {result}")
```

### 2. Importing a Library with an Alias (as)
It is common practice (especially in data science) to import a library and give it a short "nickname" (alias) to save typing.

```python
# Import Pandas and name it 'pd' (this is an industry standard convention)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Now we can use 'pd' instead of typing out 'pandas'
data = {'Name': ['Alice', 'Bob'], 'Age': [25, 30]}
df = pd.DataFrame(data)
print(df)
```

### 3. Importing Only a Specific Piece (from ... import)
If you only need one small tool from a large toolbox, you can import only that specific function or class, saving memory.

```python
# From the random library, only bring in the 'randint' tool
from random import randint

# Use 'randint' directly without referencing the library name
random_number = randint(1, 100)
print(f"Your random number is: {random_number}")
```
